```python
import os
```

```python
import re

# Example: Jul27_Scenario_3_1_2026-07-27_10-58-52
# scenario_number = "3_1", year = 2026
SCENARIO_PATTERN = re.compile(r"Scenario_(?P<number>\d+_\d+)_(?P<year>\d{4})")

def get_scenario_number(scenario):
    match = SCENARIO_PATTERN.search(str(scenario))
    return match.group("number") if match else None

def get_year(scenario):
    match = SCENARIO_PATTERN.search(str(scenario))
    return int(match.group("year")) if match else None
```

```python
Path_Rosbags = "/Users/rashid/data/DS/SR/v6/Jul27"
# Loop through every folder and collect every MCAP file.

rosbags_available = []
for root, dirs, files in os.walk(Path_Rosbags):
    dirs.sort()
    files.sort()
    for file in files:
        if file.lower().endswith(".mcap"):
            rosbag_file_path = os.path.join(root, file)
            scenario_name = os.path.basename(root)

            data__ = {
                # "name": file,
                "scenario": scenario_name,
                "scenario_ID": get_scenario_number(scenario_name),
                # "year": get_year(scenario_name),
                "path": rosbag_file_path,
                "size (MB)": os.path.getsize(rosbag_file_path) / (1024 * 1024),
                "last_modified": os.path.getmtime(rosbag_file_path),
                "created": os.path.getctime(rosbag_file_path),
            }
            rosbags_available.append(data__)

            print(
                f"Found: scenario={data__['scenario_ID']}, file={rosbag_file_path}"
            )
```

```python
import pandas as pd
df_rosbags = pd.DataFrame(rosbags_available)
df_rosbags
```
